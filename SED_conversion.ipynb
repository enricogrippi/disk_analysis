```python
# extract_sed_simple.py
import numpy as np

def extract_sed_columns(input_file, output_file):
    """Extract exactly 5 columns: Wavelength, Obs.Flux, Obs.Error, Flux, Error"""
    
    with open(input_file, 'r') as f:
        lines = f.readlines()
    
    # Find data lines (skip headers)
    data_lines = []
    for line in lines:
        line = line.strip()
        if line and not line.startswith('#') and not line.startswith('='):
            data_lines.append(line)
    
    print(f"Found {len(data_lines)} data lines")
    
    # Extract the 5 columns we need
    extracted_data = []
    for line in data_lines:
        parts = line.split()
        
        # Columns we want:
        # Column 3: Wavelength (A)
        # Column 5: Obs.Flux 
        # Column 6: Obs.Error
        # Column 7: Flux
        # Column 8: Error
        
        if len(parts) >= 9:
            wavelength_A = float(parts[1])    # Column 3 (index 2)
            wavelength_micron = wavelength_A / 10000.0  # Convert to microns
            obs_flux = parts[2]      # Column 5  
            obs_error = parts[3]     # Column 6
            flux = parts[4]          # Column 7
            error = parts[5]         # Column 8

            # Convert wavelength back to string for writing
            wavelength_str = f"{wavelength_micron:.4f}"
            extracted_data.append([wavelength_str, obs_flux, obs_error, flux, error])
    
    # Save to file (keep as strings, no conversion)
    with open(output_file, 'w') as f:
        for row in extracted_data:
            f.write(' '.join(row) + '\n')
    
    print(f"Created {output_file} with {len(extracted_data)} data points")
    print("Columns: Wavelength(A) Obs.Flux Obs.Error Flux Error")
    
    # Show first 5 lines
    print("\nFirst 5 lines:")
    for i in range(min(5, len(extracted_data))):
        print(f"  {' '.join(extracted_data[i])}")

# Run it
extract_sed_columns("wray_SED.dat", "sed_data.txt")
```
